```python
import pyreadstat
import pandas as pd

# 1) 데이터 파일 로드 (파일명이 본인 드라이브에 있는 이름과 일치하는지 확인하세요)
# 영양조사(식품섭취조사) 파일과 검진조사 파일 로드
df_ntr, meta_ntr = pyreadstat.read_sav(f'/content/drive/MyDrive/hdac_data/HN24_ALL.sav')
df_ex, meta_ex = pyreadstat.read_sas7bdat(f'/content/drive/MyDrive/hdac_data/hn24_24rc.sas7bdat')

# 2) ID 기준으로 두 파일 병합
df_merged = pd.merge(df_ntr, df_ex, on='ID', how='inner')

# 3) 분석 대상 선정 단계별 인원수 출력 (요구사항 충족)
print("=== [분석 대상 선정 단계별 인원수 확인] ===")
N0 = len(df_merged)
print(f"1단계 (전체 모집단/병합 데이터): {N0:,} 명")

# 필수 변수 리스트 (변수 정의표 기준, eat_out은 실제 코드북의 외식빈도 원변수명으로 변경하세요)
required_cols = ['eat_out', 'HE_BMI', 'region', 'sex', 'age', 'edu', 'ho_incm', 'kstrata', 'psu', 'wt_ntr']

# 결측치 제거 및 최종 인원 계산
df_filtered = df_merged.dropna(subset=required_cols)
N_filtered = len(df_filtered)
excluded_count = N0 - N_filtered

print(f"2~3단계 (결측치 및 비해당 제외 인원): {excluded_count:,} 명")
print(f"4단계 (최종 분석 대상자 수, N_final): {N_filtered:,} 명")
print("============================================")
```

```python
print("\n=== [주요 변수 빈도표 및 요약 통계 확인] ===")

# 1. 성별 (sex) 빈도표
print("\n[1] 성별 (sex) 빈도표")
print(df_filtered['sex'].value_counts(dropna=False))

# 2. 지역 (region) 빈도표
print("\n[2] 지역 (region) 빈도표")
print(df_filtered['region'].value_counts(dropna=False))

# 3. 교육수준 (edu) 빈도표
print("\n[3] 교육수준 (edu) 빈도표")
print(df_filtered['edu'].value_counts(dropna=False))

# 4. 소득 4분위수 (ho_incm) 빈도표
print("\n[4] 소득 4분위수 (ho_incm) 빈도표")
print(df_filtered['ho_incm'].value_counts(dropna=False))

# 5. 나이 (age) 요약 통계 (연속형 변수)
print("\n[5] 나이 (age) 요약 통계")
print(df_filtered['age'].describe())

# 6. 외식 빈도 빈도표 (원변수명을 'eat_out' 대신 실제 이름으로 확인)
print("\n[6] 외식 빈도 빈도표")
if 'eat_out' in df_filtered.columns:
    print(df_filtered['eat_out'].value_counts(dropna=False))
else:
    print("-> 'eat_out' 변수명을 실제 코드북의 원변수명으로 수정해 주세요!")
```
